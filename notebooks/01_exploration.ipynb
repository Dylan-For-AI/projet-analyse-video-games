## 11.Analyse de dates
On voit ici que `release_date` est du texte. Environ 1 590 valeurs ne sont pas de vraies dates comme par exemple TBA ou coming soon et 3 179 sont manquantes.

```python
dates = pd.to_datetime(df["release_date"], format="%b %d, %Y", errors="coerce")
invalides = df.loc[dates.isna() & df["release_date"].notna(), "release_date"]
print(len(invalides))
invalides.value_counts().head(10)
```
